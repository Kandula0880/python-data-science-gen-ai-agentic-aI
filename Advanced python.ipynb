```python
a = 10 
def something():
    print('in function',a)
something()
print('out function',a)
```

```python
a = 10 
def something():
    global a
    b = 15 # 15 is converted to local when user assigned global a
    print('in function',b)
    print('global variable',a)
something()
print('out function',a)
```

```python
a = 20 
def something():
    global a
    a = 15 # we refered local to global 
    print('in function',a)
    a = 1
something()
print('out function',a)
```
